```python
import os
if not "monosynaptic.py" in os.listdir():
    os.chdir("..")

from cross_correlogram import CrossCorrelogram, CrossCorrelogramSelection, CrossCorrelogramParameters
from cross_correlogram import ShuffleCrossCorrelogram, ShuffleCrossCorrelogramSelection, ShuffleCrossCorrelogramParameters
from spyglass.spikesorting.analysis.v1.group import SortedSpikesGroup
```

# Specify data

```python
nwb_file_name = 'mango20211129_.nwb'
interval_list_name = '15_s8'
sorted_spikes_group = 'mPFC_targeted_raw data valid times no premaze no home_v2'

key = {
    "nwb_file_name": nwb_file_name,
    "sorted_spikes_group_name": sorted_spikes_group,
}

sort_key = (SortedSpikesGroup() & key).fetch1("KEY")
```

```python
corr_key = {
    **sort_key,
    "interval_list_name": interval_list_name,
    "cross_corr_params_name": "monosynaptic",
}
CrossCorrelogramSelection().insert1(corr_key, skip_duplicates=True)
```

```python
CrossCorrelogram().populate(corr_key) # ~1min
CrossCorrelogram & corr_key
```

# Shuffle Dist

```python
shuffle_key = {
    **corr_key,
    "shuffle_params_name": "monosynaptic",
}

ShuffleCrossCorrelogramSelection().insert1(shuffle_key, skip_duplicates=True)
ShuffleCrossCorrelogramSelection()
```

```python
ShuffleCrossCorrelogram().populate(shuffle_key) # ~2min for 150 cells, 1 epoch, 1000 shuffles
                                                # ~1.5 hours for 500 cells, 5 epochs, 1000 shuffles
ShuffleCrossCorrelogram()
```
